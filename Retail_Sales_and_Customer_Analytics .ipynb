```python
branch_sales = df.groupby("Branch")["Total_Sales"].sum().sort_values(ascending=False)

branch_sales
```

```python
branch_sales_df = branch_sales.reset_index()
branch_sales_df.columns = ["Branch", "Total_Sales"]

branch_sales_df
```

```python
branch_performance = df.groupby("Branch").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Total_Profit=("Profit", "sum")
).sort_values("Total_Sales", ascending=False)

branch_performance
```

```python
df["Sales_Date"] = pd.to_datetime(df["Sales_Date"])

monthly_sales = df.groupby(
    df["Sales_Date"].dt.to_period("M")
)["Total_Sales"].sum()

monthly_sales
```

```python
quarterly_sales = df.groupby(
    df["Sales_Date"].dt.to_period("Q")
)["Total_Sales"].sum()

quarterly_sales
```
