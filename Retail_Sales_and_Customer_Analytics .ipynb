```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# Convert the correct date column to datetime
df["Sales_Date"] = pd.to_datetime(df["Sales_Date"])

# Group total sales by month in chronological order
monthly_sales = (
    df.groupby(df["Sales_Date"].dt.to_period("M"))["Total_Sales"]
    .sum()
    .sort_index()
)

# Format month labels
month_labels = monthly_sales.index.strftime("%b %Y")

fig, ax = plt.subplots(figsize=(12, 7))

# Create the line chart
ax.plot(
    range(len(monthly_sales)),
    monthly_sales.values,
    color="blue",
    marker="o",
    linewidth=2,
    markersize=6,
    label="Monthly Sales"
)

# Add sales values above each point
for i, value in enumerate(monthly_sales.values):
    ax.annotate(
        f"{value:,.0f}",
        (i, value),
        textcoords="offset points",
        xytext=(0, 8),
        ha="center",
        fontsize=8
    )

ax.set_title(
    "Monthly Sales Trends",
    fontsize=15,
    fontweight="bold"
)
ax.set_xlabel("Month")
ax.set_ylabel("Total Sales")

ax.set_xticks(range(len(month_labels)))
ax.set_xticklabels(month_labels, rotation=45, ha="right")

ax.yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"{x:,.0f}")
)

ax.grid(True, linestyle="--", alpha=0.4)
ax.legend(loc="best")

# Leave space for the figure caption
fig.subplots_adjust(bottom=0.27)

# Caption at the bottom-right
fig.text(
    0.98, 0.03,
    "Fig: 27 — Monthly Sales Trends",
    ha="right",
    va="bottom",
    fontsize=10,
    fontstyle="italic"
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Use the existing dataset already loaded in df
# Standardize category names to combine duplicates
df["Category"] = (
    df["Category"].astype("string").str.strip().str.title()
)

df["Total_Sales"] = pd.to_numeric(
    df["Total_Sales"], errors="coerce"
)

data = df.dropna(subset=["Category", "Total_Sales"]).copy()

# Combine duplicate categories and calculate total sales
category_sales = (
    data.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

total_sales = category_sales.sum()
percentages = category_sales / total_sales * 100

# Display exact results from your dataset
summary = pd.DataFrame({
    "Product Category": category_sales.index,
    "Total Sales": category_sales.values.round(2),
    "Percentage (%)": percentages.values.round(2)
})

print("QUESTION 28: PRODUCT CATEGORY SALES")
display(summary)

# Draw bar chart
colors = ["royalblue", "darkorange", "forestgreen", "crimson"]

fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.bar(
    category_sales.index,
    category_sales.values,
    color=colors[:len(category_sales)],
    edgecolor="black",
    width=0.65
)

# Add values and percentages above bars
for bar, value, pct in zip(
    bars, category_sales.values, percentages.values
):
    ax.annotate(
        f"{value:,.2f}\n({pct:.1f}%)",
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, 6),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=9
    )

ax.set_title("Product Category Sales", fontsize=15, fontweight="bold")
ax.set_xlabel("Product Category")
ax.set_ylabel("Total Sales")
ax.grid(axis="y", linestyle="--", alpha=0.3)
ax.set_axisbelow(True)
ax.set_ylim(0, category_sales.max() * 1.18)

# Legend
legend_items = [
    Patch(facecolor=colors[i], edgecolor="black", label=category)
    for i, category in enumerate(category_sales.index)
]
ax.legend(handles=legend_items, title="Product Categories",
          loc="upper right")

# Figure caption
fig.text(
    0.98, 0.01,
    "Fig: 28 — Product Category Sales",
    ha="right", fontsize=9, style="italic"
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# 1. Load the dataset
df = pd.read_csv("retail_sales_project_dataset (1).csv")

# 2. Convert Customer_Age to numeric and remove invalid ages
age = pd.to_numeric(df["Customer_Age"], errors="coerce")
valid_age = age[age.between(18, 64)].dropna()

# 3. Define age groups
bins = [18, 25, 35, 45, 55, 65]
labels = ["18–24", "25–34", "35–44", "45–54", "55–64"]

age_groups = pd.cut(
    valid_age,
    bins=bins,
    labels=labels,
    right=False
)

# 4. Calculate counts and percentages
counts = age_groups.value_counts().reindex(labels, fill_value=0)
total = counts.sum()
percentages = (counts / total * 100) if total > 0 else counts.astype(float)

# Display the summary table
summary = pd.DataFrame({
    "Age Range": labels,
    "Customer Count": counts.values,
    "Percentage (%)": percentages.values.round(2)
})

print("CUSTOMER AGE DISTRIBUTION")
print(summary.to_string(index=False))
print(f"\nTotal customers with valid ages (18–64): {total}")
print(f"Missing ages: {age.isna().sum()}")
print(f"Ages below 18: {(age < 18).sum()}")
print(f"Ages above 64: {(age > 64).sum()}")

# 5. Plot the bar chart
colors = ["royalblue", "darkorange", "forestgreen", "crimson", "purple"]

fig, ax = plt.subplots(figsize=(11, 6))

bars = ax.bar(
    labels,
    counts.values,
    color=colors,
    edgecolor="black",
    linewidth=0.7
)

# Add count and percentage labels above each bar
for bar, count, percentage in zip(
    bars, counts.values, percentages.values
):
    ax.annotate(
        f"{count:,}\n({percentage:.2f}%)",
        xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, 5),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

# 6. Titles, axes, legend and styling
ax.set_title("Customer Age Distribution", fontsize=15, fontweight="bold")
ax.set_xlabel("Customer Age Range (years)", fontsize=12)
ax.set_ylabel("Number of Customers", fontsize=12)

ax.set_ylim(0, max(counts.max() * 1.18, 1))
ax.grid(axis="y", linestyle="--", alpha=0.35)
ax.set_axisbelow(True)

legend_items = [
    Patch(facecolor=colors[i], edgecolor="black", label=labels[i])
    for i in range(len(labels))
]

ax.legend(
    handles=legend_items,
    title="Age Groups",
    loc="upper right"
)

# Figure caption at the bottom-right
fig.text(
    0.99, 0.01,
    "Fig: 30 — Customer Age Distribution",
    ha="right",
    va="bottom",
    fontsize=9,
    style="italic"
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.patches import Patch

# Group total sales by discount percentage
sales_by_discount = (
    df.groupby("Discount")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

# Calculate percentage contribution to total sales
percentages = (
    sales_by_discount / sales_by_discount.sum()
) * 100

# Different colors for each discount level
colors = [
    "steelblue", "darkorange", "forestgreen",
    "crimson", "purple", "gold", "teal"
]
bar_colors = [
    colors[i % len(colors)]
    for i in range(len(sales_by_discount))
]

# Create chart
fig, ax = plt.subplots(figsize=(12, 7))

bars = ax.bar(
    sales_by_discount.index.astype(str) + "%",
    sales_by_discount.values,
    color=bar_colors,
    edgecolor="black",
    width=0.7
)

# Display sales values and percentages above bars
for i, bar in enumerate(bars):
    ax.annotate(
        f"{sales_by_discount.iloc[i]:,.2f}\n"
        f"({percentages.iloc[i]:.1f}%)",
        xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, 5),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=9
    )

# Titles and labels
ax.set_title(
    "Relationship Between Discounts and Total Sales",
    fontsize=14,
    fontweight="bold",
    pad=15
)
ax.set_xlabel("Discount")
ax.set_ylabel("Total Sales")

ax.yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"{x:,.0f}")
)

ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.set_axisbelow(True)
ax.margins(y=0.20)

# Legend INSIDE the chart at the top-right
legend_items = [
    Patch(
        facecolor=bar_colors[i],
        edgecolor="black",
        label=f"{discount}% discount"
    )
    for i, discount in enumerate(sales_by_discount.index)
]

ax.legend(
    handles=legend_items,
    title="Discount Levels",
    loc="upper right",
    frameon=True,
    framealpha=0.95,
    fontsize=9
)

# Leave space for figure caption
fig.subplots_adjust(bottom=0.20)

# Figure number and title at bottom-right
fig.text(
    0.98, 0.04,
    "Fig: 31 — Relationship Between Discounts and Total Sales",
    ha="right",
    va="bottom",
    fontsize=10,
    fontstyle="italic"
)

plt.show()

# Print summary for interpretation
summary = pd.DataFrame({
    "Total Sales": sales_by_discount,
    "Sales Contribution (%)": percentages.round(1)
})

print("Total Sales by Discount Level")
print(summary.to_string())
print(f"\nTotal sales: {sales_by_discount.sum():,.2f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Prepare profit data
profit_df = df[["Branch", "Profit"]].copy()

profit_df["Profit"] = pd.to_numeric(
    profit_df["Profit"], errors="coerce"
)

profit_df = profit_df.dropna(subset=["Branch", "Profit"])

# Get branches and their profit values
branches = sorted(profit_df["Branch"].unique())

profit_data = [
    profit_df.loc[profit_df["Branch"] == branch, "Profit"].values
    for branch in branches
]

# Different colors for branches
colors = [
    "royalblue", "darkorange", "forestgreen", "crimson",
    "purple", "gold", "teal", "deeppink", "brown", "gray"
]

branch_colors = [
    colors[i % len(colors)] for i in range(len(branches))
]

# Create chart
fig, ax = plt.subplots(figsize=(12, 7))

box = ax.boxplot(
    profit_data,
    patch_artist=True,
    tick_labels=branches,  # Corrected argument
    medianprops=dict(color="black", linewidth=2),
    whiskerprops=dict(color="black"),
    capprops=dict(color="black"),
    flierprops=dict(
        marker="o",
        markerfacecolor="gray",
        markeredgecolor="black",
        markersize=4,
        alpha=0.5
    )
)

# Apply colors
for patch, color in zip(box["boxes"], branch_colors):
    patch.set_facecolor(color)
    patch.set_edgecolor("black")
    patch.set_alpha(0.85)

# Titles and labels
ax.set_title(
    "Branch Profitability Distribution",
    fontsize=15,
    fontweight="bold",
    pad=15
)

ax.set_xlabel("Branch", fontsize=11)
ax.set_ylabel("Profit", fontsize=11)

ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.set_axisbelow(True)

# Legend inside the chart at the top-right
legend_items = [
    Patch(
        facecolor=branch_colors[i],
        edgecolor="black",
        label=str(branches[i])
    )
    for i in range(len(branches))
]

ax.legend(
    handles=legend_items,
    title="Branches",
    loc="upper right",
    frameon=True,
    framealpha=0.95,
    fontsize=9
)

# Space for figure caption
fig.subplots_adjust(bottom=0.20)

# Figure caption at the bottom-right
fig.text(
    0.98, 0.04,
    "Fig: 32 — Branch Profitability Using Boxplots",
    ha="right",
    va="bottom",
    fontsize=10,
    fontstyle="italic"
)

plt.show()

# Print branch profitability summary
print("Profit Summary by Branch")
print(
    profit_df.groupby("Branch")["Profit"]
    .describe()
    .round(2)
    .to_string()
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

d = df.dropna(subset=["Category", "Return_Status"]).copy()
d["Category"] = d["Category"].str.strip().str.title()
d["Returned"] = d["Return_Status"].str.strip().str.lower().eq("returned").astype(int)

table = d.groupby("Category")["Returned"].agg(["sum", "count"])
table.columns = ["Returned", "Total"]
table["Return Rate (%)"] = (table["Returned"] / table["Total"] * 100).round(1)

print("PRODUCT RETURN RATES BY CATEGORY")
print(table.to_string())

table = table.sort_values("Return Rate (%)", ascending=False)
colors = ["royalblue", "darkorange", "forestgreen", "crimson"]
fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(table.index, table["Return Rate (%)"],
              color=colors[:len(table)])

for b, v in zip(bars, table["Return Rate (%)"]):
    ax.text(b.get_x() + b.get_width()/2, v + 0.5, f"{v:.1f}%",
            ha="center", fontsize=9)

ax.set_title("Product Return Rates by Category")
ax.set_xlabel("Product Category")
ax.set_ylabel("Return Rate (%)")
ax.legend(handles=[Patch(color=colors[i], label=c)
                   for i, c in enumerate(table.index)],
          title="Category", loc="upper right")
ax.grid(axis="y", linestyle="--", alpha=.3)
ax.set_axisbelow(True)
fig.text(.98, .01, "Fig: 34 — Product Return Rates by Category",
         ha="right", fontsize=9, style="italic")
plt.tight_layout(rect=[0, .05, 1, 1])
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Select numerical columns available in your dataset
cols = ["Customer_Age", "Unit_Price", "Quantity", "Discount",
        "Tax", "Total_Sales", "Cost", "Profit", "Stock_Level",
        "Customer_Rating", "Satisfaction"]

cols = [c for c in cols if c in df.columns]
data = df[cols].apply(pd.to_numeric, errors="coerce")
corr = data.corr()

# Display correlation table
print("CORRELATION MATRIX")
print(corr.round(2).to_string())

# Plot heatmap using Matplotlib
fig, ax = plt.subplots(figsize=(11, 8))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)

ax.set_xticks(range(len(cols)), labels=cols, rotation=45, ha="right")
ax.set_yticks(range(len(cols)), labels=cols)

for i in range(len(cols)):
    for j in range(len(cols)):
        if pd.notna(corr.iloc[i, j]):
            ax.text(j, i, f"{corr.iloc[i, j]:.2f}",
                    ha="center", va="center", fontsize=8)

fig.colorbar(im, ax=ax, label="Correlation Coefficient")
ax.set_title("Correlation Matrix of Numerical Business Variables",
             fontweight="bold", pad=15)

fig.text(.99, .01, "Fig: 35 — Correlation Matrix",
         ha="right", fontsize=9, style="italic")

plt.tight_layout(rect=[0, .04, 1, 1])
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Calculate total sales by product
d = df.copy()
d["Total_Sales"] = pd.to_numeric(d["Total_Sales"], errors="coerce")
sales = d.dropna(subset=["Product_Name", "Total_Sales"])
sales = sales.groupby("Product_Name")["Total_Sales"].sum()
sales = sales.sort_values(ascending=False)

colors = ["steelblue", "darkorange", "green", "red", "purple",
          "brown", "pink", "gray", "olive", "teal"]

def draw_chart(data, title, fig_no):
    pct = data / data.sum() * 100

    # Data table
    table = pd.DataFrame({
        "Total Sales": data.round(2),
        "Percentage (%)": pct.round(1)
    })
    print("\n" + title.upper())
    print(table.to_string())

    # Bar chart
    fig, ax = plt.subplots(figsize=(11, 6))
    bars = ax.bar(data.index, data.values, color=colors[:len(data)],
                  edgecolor="black")

    for i, b in enumerate(bars):
        ax.text(b.get_x() + b.get_width()/2, b.get_height(),
                f"{data.iloc[i]:,.2f}\n({pct.iloc[i]:.1f}%)",
                ha="center", va="bottom", fontsize=8)

    ax.set_title(title, fontweight="bold")
    ax.set_xlabel("Product")
    ax.set_ylabel("Total Sales")
    ax.grid(axis="y", linestyle="--", alpha=.3)
    ax.set_axisbelow(True)
    ax.margins(y=.23)
    plt.xticks(rotation=35, ha="right")

    ax.legend(handles=[
        Patch(color=colors[i], label=str(name))
        for i, name in enumerate(data.index)
    ], title="Products", loc="upper right", fontsize=8)

    fig.text(.98, .01, f"Fig: {fig_no} — {title}",
             ha="right", fontsize=9, style="italic")
    plt.tight_layout(rect=[0, .06, 1, 1])
    plt.show()

# Chart 1: Top 10 products
draw_chart(sales.head(10), "Top 10 Products by Total Sales", "36A")

# Chart 2: Bottom 5 products
draw_chart(sales.tail(5).sort_values(),
           "Bottom 5 Products by Total Sales", "36B")
```
