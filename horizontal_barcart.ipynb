HORIZONTAL BARCHART

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
```

```python
df=pd.read_excel('datasets/nyeri.xlsx')
df.head(5)
```

```python
list(df.columns)
```

```python
df['student_category'].value_counts()
```

```python
string_columns = df.select_dtypes(include=['object', 'string']).columns

df[string_columns] = df[string_columns].astype('string').apply(
    lambda x: x.str.strip().str.upper()
)
```

```python
category_count=df['student_category'].value_counts()
plt.barh(
    category_count.index,
    category_count.values,
    color=['#3399ff','#ff99ff','#cc66ff','#00cc99'],
    height=0.8
)

plt.title(
    'Distribution of Student By Category',
    color='#000099',
    fontsize=13,
    fontstyle='italic',
    fontweight='bold'
)

plt.xlabel(
    'No Of Students'
)

plt.ylabel(
    'Student Category'
)

plt.figtext(1,0,'Fig:8.0')

colors=[
    Patch(facecolor='#3399ff',label='CBS'),
    Patch(facecolor='#ff99ff',label='KNEC'),
    Patch(facecolor='#cc66ff',label='CBET'),
    Patch(facecolor='#00cc99',label='CDACC')
]

plt.legend(
    handles=colors,
    title='Category'
)

for i, value in enumerate(category_count.values):
    percentage=(value/category_count.sum())*100
    plt.text(
        value +20,
        i,
        f'{value} ({percentage: .2f}%)',
        color='red',
        fontweight='bold'
    )
```
